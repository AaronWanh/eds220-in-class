## Import libraries

```python
import os

import pandas as pd
from pandas.api.types import is_string_dtype, is_numeric_dtype
import geopandas as gpd
import matplotlib.pyplot as plt


# Display all columns when looking at dataframes 
pd.set_option("display.max_columns", None)
```

```python
# Create points from latitude and longitude 
points = gpd.points_from_xy(power_plants['longitude'],
                            power_plants['latitude'])

# Create geodataframe
power_plants = gpd.GeoDataFrame(power_plants,         # Data
                                geometry = points,    # Specify geometry column 
                                crs = 'EPSG:4269')    # Specify CRS
```

```python
power_plants.geometry
```

With the `geometry` column and CRS, we can plot our dataset:

```python
power_plants.plot()
```

## f-strings

So far, we have printed variables by passing several arguments to the `print()` function. This means that we write commas between every string and variable we want to print, and then the `print()` function joins these into a single string, with a space between each one. For example:

```python
print('CRS', power_plants.crs)
```

Another popular way of mixing strings and variables in print statements is by creating and *f-string* which stnads for "formatted string", The simplest syntax for an f-string is:

```python
f" some text {replace}"
```

where `replace` can be a variable, an expression, or a function or method call. For example:

```python
# Explore CRS 
print(f"ellipsoid: {power_plants.crs.ellipsoid}")
print(f"datum: {power_plants.crs.datum}")
```

We just created a string replacing the value inside the curly brackets `{}`.

One of the advantages of using f-strings is that they offer customization for formatting the output:

```python
# Any expression can go inside the curly brackets
# :, adds thousands separators
print(f"The data has {len(power_plants):,} power plants in {power_plants['state'].nunique()} states and territories.")

# :.2f rounds to 2 decimals 
print(f"Average installed capacity: {power_plants['install_mw'].mean():.2f} MW")
```

## `for` loops

The `region`, `division`, and `statefp` columns contain codes. The US Census Bureau groups the states into four regions (for example, the West), which are further split into nine divisions (for example, the Pacific division). Each state also has a FIPS (Federal Information Processing Standards) code, a numeric code that identifies it. For example, California's FIPS code is 06. We will use these codes to select the states we need. 

It can be easier to work with the coes as numbers instead of strings, so let's update the corresponding columns in the states geodataframe. We start by checking the data type of the `region`, `division`, and `statefp` columns:

```python
code_cols = ['region', 'division', 'statefp']

# Check whether codes columns are strings 
for column in code_cols:
    print(f"{column} is string dtype? {is_string_dtype(states[column])}")
```

Remember `for ` loops execute a block of code a fix ednumber of times, iterating over a set of obsejcts. In this case, we iterate over the list of column names `code_cols = ['region', 'division', 'statefp']`.

## `asset`

Next, we update the data type of the code columns to be integers. This time, we check the data type of the column using the `is_numeric_dtype()` function inside an `assert` statement:


```python
# Update code columns into integers
for column in code_cols:
    states[column] = states[column].astype('int')
    assert is_numeric_dtype(states[column])  # Check conversion
```

The `assert` keywrod does nothing if the expression next to it evaluates to `True`. If it evaluates to `False`, it raises an `AssertionError` exception and stops your code from running any further. For example, 

```python
# Does nothing if statement is True
assert 2+2 == 4

# Raises an error if statement is False
assert 2+2 == 3
```
