The females may have been pregnant leading to a larger average weight for female snowshow hares. 

```python
import pandas as pd
import numpy as np

# Read snowshoe hare dataset
hares = pd.read_csv('data/55_Hare_Data_2012.txt')

# cleaning the data to reduce the amount of variables into a new column
hares['simple_sex'] = hares['sex']
    .str.upper()
    .str.strip(" ")
    .replace({
    "PF" : "F?"
})

hares.loc[hares['simple_sex'] == "?", 'simple_sex'] = np.nan

# View the new cleaned dataset
hares['simple_sex'].value_counts(dropna=False)

# Calculate the average weight for female and male hares separately 
hares.groupby('simple_sex')['weight'].mean()
```
